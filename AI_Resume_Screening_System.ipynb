```python

# Sample Resume
resume = """
Python, SQL, Pandas, Data Analysis,
Machine Learning, Excel, Data Visualization
"""

# Job Description
job_description = """
We are looking for a candidate with Python,
SQL, Machine Learning, Pandas,
Power BI and Communication skills.
"""

print("Resume and Job Description loaded successfully!")
```

```python

# Skills required for the job
skills = [
    "Python",
    "SQL",
    "Pandas",
    "Machine Learning",
    "Power BI",
    "Communication"
]

# Find matched and missing skills
matched_skills = []
missing_skills = []

for skill in skills:
    if skill.lower() in resume.lower():
        matched_skills.append(skill)
    else:
        missing_skills.append(skill)

print("Matched Skills:", matched_skills)
print("Missing Skills:", missing_skills)
```

```python

total_skills = len(skills)
matched_count = len(matched_skills)

match_percentage = (matched_count / total_skills) * 100

print("Resume Match Percentage:", round(match_percentage, 2), "%")
```

```python

print("===== AI RESUME SCREENING REPORT =====")

print("Total Required Skills:", len(skills))
print("Matched Skills:", len(matched_skills))
print("Missing Skills:", len(missing_skills))

print("\nMatched Skills:")
for skill in matched_skills:
    print("-", skill)

print("\nMissing Skills:")
for skill in missing_skills:
    print("-", skill)

print("\nResume Match Percentage:", round(match_percentage, 2), "%")
print("======================================")
```

```python

from pypdf import PdfReader

pdf_file = list(uploaded.keys())[0]
reader = PdfReader(pdf_file)

resume_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        resume_text += text + "\n"

print(resume_text)
```

```python

# Use the extracted resume text
resume = resume_text

# Job description
job_description = """
We are looking for a candidate with Python,
SQL, Pandas, Machine Learning, Power BI,
Data Analysis and Communication skills.
"""

# Required skills
skills = [
    "Python",
    "SQL",
    "Pandas",
    "Machine Learning",
    "Power BI",
    "Data Analysis",
    "Communication"
]

# Compare resume with job requirements
matched_skills = []
missing_skills = []

for skill in skills:
    if skill.lower() in resume.lower():
        matched_skills.append(skill)
    else:
        missing_skills.append(skill)

# Calculate match percentage
match_percentage = (
    len(matched_skills) / len(skills)
) * 100

# Display screening report
print("===== RESUME SCREENING REPORT =====")
print("Matched Skills:", matched_skills)
print("Missing Skills:", missing_skills)
print("Match Percentage:", round(match_percentage, 2), "%")
```

```python

print("\n===== FINAL SCREENING RESULT =====")

if match_percentage >= 70:
    print("Result: High Skill Match")
elif match_percentage >= 40:
    print("Result: Moderate Skill Match")
else:
    print("Result: Low Skill Match")

print("Match Score:", round(match_percentage, 2), "%")
print("Matched Skills:", len(matched_skills))
print("Missing Skills:", len(missing_skills))

print("\nSkills to Improve:")
for skill in missing_skills:
    print("-", skill)

print("=================================")
```

```python

job_description = input("Enter the Job Description: ")

skills = [
    "Python",
    "SQL",
    "Pandas",
    "Machine Learning",
    "Power BI",
    "Data Analysis",
    "Communication"
]

matched_skills = []
missing_skills = []

for skill in skills:
    if skill.lower() in resume_text.lower() and skill.lower() in job_description.lower():
        matched_skills.append(skill)
    elif skill.lower() in job_description.lower():
        missing_skills.append(skill)

total_required = len(matched_skills) + len(missing_skills)

if total_required > 0:
    match_percentage = len(matched_skills) / total_required * 100
else:
    match_percentage = 0

print("\n===== RESUME SCREENING REPORT =====")
print("Matched Skills:", matched_skills)
print("Missing Skills:", missing_skills)
print("Match Percentage:", round(match_percentage, 2), "%")
```

```python

print("\n========== AI RESUME SCREENING SYSTEM ==========")

print("\nResume Analysis Summary")
print("-----------------------")
print("Total Required Skills:", len(matched_skills) + len(missing_skills))
print("Skills Matched:", len(matched_skills))
print("Skills Missing:", len(missing_skills))
print("Overall Match Score:", round(match_percentage, 2), "%")

print("\nRecommended Skills to Learn")
print("---------------------------")

if missing_skills:
    for skill in missing_skills:
        print("-", skill)
else:
    print("All listed skills matched!")

print("\nNote: This score is based on keyword matching.")
print("It is not a final hiring decision.")
print("================================================")
```

```python

import pandas as pd

report = pd.DataFrame({
    "Skill": skills,
    "Status": [
        "Matched" if skill in matched_skills else "Missing"
        for skill in skills
    ]
})

report.loc[len(report)] = [
    "Overall Match Percentage",
    f"{match_percentage:.2f}%"
]

report.to_csv("resume_screening_report.csv", index=False)

print("Screening report saved successfully!")
```
